```python
from pandas import read_csv, DataFrame, get_dummies
data = read_csv('/content/bank.csv')
data.head(10)
data.info()
data.describe()
```

```python
# 2. Binary mapping for simple yes/no columns
data['default'] = data['default'].map({'no':0, 'yes':1})
data['housing'] = data['housing'].map({'no':0, 'yes':1})
data['loan'] = data['loan'].map({'no':0, 'yes':1})
data['y'] = data['y'].map({'no':0, 'yes':1})
data.head()
```

```python
from pandas import read_csv, DataFrame, get_dummies
# 3. Dummy encode the remaining categorical text columns
data_encoded = get_dummies(data, columns=['job', 'marital', 'education', 'contact', 'month', 'poutcome'], dtype=int)
```

```python
data.info()
```

```python
# 4. Separate Features (x) and Target (y)
x = data_encoded.drop('y', axis=1)
y = data_encoded['y']
print(x.shape)
print(y.shape)
```

```python
# 5. Scale the features so large numbers don't dominate
from sklearn.preprocessing import StandardScaler
x_scaled = StandardScaler().fit_transform(x)
DataFrame(x_scaled)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x_scaled, y, test_size=0.10, random_state=10)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
# 7. Balance ONLY the training data using SMOTE
from imblearn.over_sampling import SMOTE
x_train, y_train = SMOTE(random_state=20).fit_resample(x_train, y_train)
```

```python
# Build and Tune Model
dt = DecisionTreeClassifier(criterion='entropy')
dt_params = {'max_depth': [1-8]}

grid_search_dt = GridSearchCV(estimator=dt, param_grid=dt_params, scoring='f1', cv=5)
grid_search_dt.fit(x_train, y_train)
y_pred_dt = grid_search_dt.predict(x_test)

# Print Results
print("--- Decision Tree ---")
print("Best DT Parameters:", grid_search_dt.best_params_)
print(confusion_matrix(y_test, y_pred_dt))
print("Accuracy: ", round(accuracy_score(y_test, y_pred_dt) * 100, 2), "%")
print("Precision:", round(precision_score(y_test, y_pred_dt) * 100, 2), "%")
print("Recall:   ", round(recall_score(y_test, y_pred_dt) * 100, 2), "%")
print("F1 Score: ", round(f1_score(y_test, y_pred_dt) * 100, 2), "%\n")
```

```python
Step 3: Random Forest
This tunes the Random Forest and prints out the metrics manually.
# Build and Tune Model
rf = RandomForestClassifier(criterion='entropy', random_state=10)
rf_params = {'n_estimators': [8-12]}

grid_search_rf = GridSearchCV(estimator=rf, param_grid=rf_params, scoring='f1', cv=5)
grid_search_rf.fit(x_train, y_train)
y_pred_rf = grid_search_rf.predict(x_test)

# Print Results
print("--- Random Forest ---")
print("Best RF Parameters:", grid_search_rf.best_params_)
print(confusion_matrix(y_test, y_pred_rf))
print("Accuracy: ", round(accuracy_score(y_test, y_pred_rf) * 100, 2), "%")
print("Precision:", round(precision_score(y_test, y_pred_rf) * 100, 2), "%")
print("Recall:   ", round(recall_score(y_test, y_pred_rf) * 100, 2), "%")
print("F1 Score: ", round(f1_score(y_test, y_pred_rf) * 100, 2), "%\n")
```
